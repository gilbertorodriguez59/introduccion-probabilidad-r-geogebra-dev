# Capítulo 2. Conjuntos, espacios muestrales y eventos
Cuaderno en R para representar espacios muestrales y operaciones entre eventos.

```r
Omega <- 1:12
A <- Omega[Omega %% 2 == 0]
B <- Omega[Omega > 6]
list(Omega=Omega, A=A, B=B, union=union(A,B), interseccion=intersect(A,B), A_complemento=setdiff(Omega,A))
```
