# Capítulo 16 — Esperanza conjunta, covarianza y correlación
Cuaderno en R para estudiar momentos conjuntos, covarianza, correlación y dependencia no lineal.

```r
P <- matrix(c(.20,.10,.05,.15,.20,.10,.05,.10,.05),3,byrow=TRUE)
x <- 0:2; y <- 0:2
px <- rowSums(P); py <- colSums(P)
EX <- sum(x*px); EY <- sum(y*py)
EXY <- sum(outer(x,y,'*')*P)
VX <- sum(x^2*px)-EX^2; VY <- sum(y^2*py)-EY^2
CXY <- EXY-EX*EY
c(EX=EX,EY=EY,EXY=EXY,cov=CXY,rho=CXY/sqrt(VX*VY))
```

## Correlación controlada

```r
set.seed(2026)
n <- 5000; rho <- .8
z1 <- rnorm(n); z2 <- rnorm(n)
X <- z1; Y <- rho*z1+sqrt(1-rho^2)*z2
cor(X,Y)
plot(X,Y,pch=16,cex=.35)
```

## Dependencia no lineal con correlación aproximadamente cero

```r
set.seed(2026)
X <- runif(50000,-1,1); Y <- X^2
cor(X,Y)
plot(X,Y,pch=16,cex=.25)
```
