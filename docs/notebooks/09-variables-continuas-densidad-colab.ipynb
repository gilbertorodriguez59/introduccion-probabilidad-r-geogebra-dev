## Densidad y normalización
Trabajaremos con $f_X(x)=x/2$ para $0\le x\le2$.


```r
f <- function(x) ifelse(x >= 0 & x <= 2, x / 2, 0)
integrate(f, 0, 2)
integrate(f, 0.5, 1.5)
```

## CDF
Para esta densidad, $F_X(x)=x^2/4$ en $[0,2]$.


```r
F <- function(x) {
  if (x <= 0) return(0)
  if (x >= 2) return(1)
  x^2 / 4
}
sapply(c(-1,0,0.5,1,1.5,2,3), F)
```
