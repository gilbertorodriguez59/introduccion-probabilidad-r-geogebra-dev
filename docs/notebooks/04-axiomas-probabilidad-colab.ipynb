# Capítulo 4. Axiomas y propiedades de la probabilidad
Cuaderno en R para verificar reglas de unión, complemento y coherencia probabilística.

```r
pA <- 0.55; pB <- 0.40; pAB <- 0.20
pUnion <- pA + pB - pAB
c(P_union=pUnion, P_ninguno=1-pUnion, P_exactamente_uno=pA+pB-2*pAB)
```

```r
inferior <- max(0, pA+pB-1)
superior <- min(pA,pB)
c(limite_inferior=inferior, limite_superior=superior, interseccion_valida=pAB>=inferior && pAB<=superior)
```
