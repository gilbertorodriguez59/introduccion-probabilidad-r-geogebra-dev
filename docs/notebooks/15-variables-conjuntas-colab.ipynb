# Capítulo 15. Variables aleatorias conjuntas
PMF conjunta, marginales, condicionales, independencia y simulación en R.

```r
P <- matrix(c(0.20,0.10,0.05,0.15,0.20,0.10,0.05,0.10,0.05),3,byrow=TRUE)
P
px <- rowSums(P); py <- colSums(P)
px; py
```

```r
P[,2]/py[2]
P_ind <- outer(px,py)
P_ind
max(abs(P-P_ind))
```

```r
set.seed(2026)
n <- 20000
u1 <- runif(n); u2 <- runif(n)
X <- sqrt(u1); Y <- X*u2
plot(X,Y,pch=16,cex=.3,xlab='X',ylab='Y',main='Soporte triangular')
abline(0,1,lty=2)
```
