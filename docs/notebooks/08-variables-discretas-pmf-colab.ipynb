## 1. Definir una PMF
Trabajaremos con una variable discreta de soporte $\{0,1,2,3\}$.


```r
x <- 0:3
p <- c(0.10, 0.25, 0.40, 0.25)
sum(p)
```

## 2. Calcular probabilidades


```r
P_ge_2 <- sum(p[x >= 2])
P_lt_2 <- sum(p[x < 2])
c(P_X_ge_2=P_ge_2, P_X_lt_2=P_lt_2)
```

## 3. PMF y CDF


```r
F <- cumsum(p)
data.frame(x=x, pmf=p, cdf=F)
```

```r
plot(x,p,type='h',lwd=5,ylim=c(0,0.5),xlab='x',ylab='p_X(x)')
points(x,p,pch=19)
```

```r
plot(x,F,type='s',ylim=c(0,1.05),xlab='x',ylab='F_X(x)')
points(x,F,pch=19)
```

## 4. Simulación


```r
set.seed(2026)
n <- 50000
muestra <- sample(x,n,replace=TRUE,prob=p)
prop.table(table(muestra))
```
