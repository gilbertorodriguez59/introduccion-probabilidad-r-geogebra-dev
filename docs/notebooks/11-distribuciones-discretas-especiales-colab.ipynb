## Bernoulli y binomial


```r
p <- 0.4
n <- 12
x <- 0:n
pmf <- dbinom(x, size=n, prob=p)
data.frame(x=x, pmf=pmf, cdf=pbinom(x, size=n, prob=p))
```

```r
plot(x, pmf, type='h', lwd=4, xlab='x', ylab='P(X=x)')
points(x, pmf, pch=19)
c(media=n*p, varianza=n*p*(1-p))
```

## Geométrica
En el libro X cuenta ensayos hasta el primer éxito; R cuenta fracasos, por lo que usamos X = Y + 1.


```r
p <- 0.3
xg <- 1:15
pg <- dgeom(xg-1, prob=p)
data.frame(x=xg, pmf=pg)
c(media=1/p, varianza=(1-p)/p^2)
```

## Binomial negativa
En el libro X cuenta ensayos totales hasta el r-ésimo éxito; R cuenta fracasos, por lo que X = Y + r.


```r
r <- 5
p <- 0.8
xnb <- r:15
pnb <- dnbinom(xnb-r, size=r, prob=p)
data.frame(x=xnb, pmf=pnb)
c(media=r/p, varianza=r*(1-p)/p^2)
```

## Simulación comparativa


```r
set.seed(2026)
b <- rbinom(50000, size=12, prob=0.4)
g <- rgeom(50000, prob=0.3)+1
nb <- rnbinom(50000, size=5, prob=0.8)+5
rbind(binomial=c(mean(b),var(b)), geometrica=c(mean(g),var(g)), binomial_negativa=c(mean(nb),var(nb)))
```
