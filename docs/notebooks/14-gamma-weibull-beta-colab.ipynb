## Distribución Gamma


```r
alpha <- 4
lambda <- 3
pgamma(2, shape=alpha, rate=lambda)
qgamma(0.95, shape=alpha, rate=lambda)
c(media=alpha/lambda, varianza=alpha/lambda^2)
```

## Distribución Weibull y supervivencia


```r
k <- 2
eta <- 1000
pweibull(1200, shape=k, scale=eta, lower.tail=FALSE)
hazard <- function(x,k,eta) (k/eta)*(x/eta)^(k-1)
hazard(c(500,1000,1500), k, eta)
```

## Distribución Beta


```r
a <- 5
b <- 2
pbeta(0.8, a, b, lower.tail=FALSE)
c(media=a/(a+b), varianza=a*b/((a+b)^2*(a+b+1)))
```

## Simulación comparativa


```r
set.seed(2026)
xg <- rgamma(50000, shape=4, rate=3)
xw <- rweibull(50000, shape=2, scale=1000)
xb <- rbeta(50000, 5, 2)
c(mean(xg), var(xg))
c(mean(xw), var(xw))
c(mean(xb), var(xb))
```
