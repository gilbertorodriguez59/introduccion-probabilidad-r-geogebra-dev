# Capítulo 7. Variable aleatoria y función de distribución acumulada
Cuaderno en R para transformar resultados, construir distribuciones y visualizar CDF.

```r
omega <- c('CC','CS','SC','SS')
X <- c(2,1,1,0)
data.frame(resultado=omega, X=X)
```

```r
x <- 0:2
p <- c(.25,.50,.25)
F <- cumsum(p)
data.frame(x=x, P_X=p, F_X=F)
plot(x,F,type='s',ylim=c(0,1),xlab='x',ylab='F_X(x)')
points(x,F,pch=19)
```
