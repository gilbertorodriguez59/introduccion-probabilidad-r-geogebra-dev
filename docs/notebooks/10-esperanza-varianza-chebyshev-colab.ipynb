## Distribución discreta
Trabajaremos con $X\in\{0,1,2,3\}$ y probabilidades $(0.70,0.20,0.08,0.02)$.


```r
x <- 0:3
p <- c(0.70, 0.20, 0.08, 0.02)
sum(p)
```

## Esperanza, segundo momento y varianza


```r
mu <- sum(x*p)
EX2 <- sum(x^2*p)
varX <- EX2 - mu^2
sdX <- sqrt(varX)
c(media=mu, segundo_momento=EX2, varianza=varX, desviacion=sdX)
```

## Esperanza de una función


```r
costo <- function(x) 5 + 3*x + 2*x^2
sum(costo(x)*p)
```

## Simulación


```r
set.seed(2026)
muestra <- sample(x, 100000, replace=TRUE, prob=p)
c(media_empirica=mean(muestra), varianza_empirica=var(muestra), desviacion_empirica=sd(muestra))
```

## Chebyshev
Comparamos la probabilidad real dentro de $\mu\pm k\sigma$ con la garantía mínima $1-1/k^2$.


```r
k <- 2
li <- mu-k*sdX
ls <- mu+k*sdX
prob_real <- sum(p[x>li & x<ls])
c(probabilidad_real=prob_real, garantia_chebyshev=1-1/k^2)
```
