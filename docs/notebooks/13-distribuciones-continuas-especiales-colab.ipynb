# Capítulo 13. Distribuciones continuas especiales
Cuaderno complementario en R: uniforme, exponencial y normal.


## Uniforme


```r
a <- 2; b <- 8
punif(6,a,b)-punif(3,a,b)
(a+b)/2
(b-a)^2/12
```

## Exponencial y propiedad sin memoria


```r
lambda <- 0.25
pexp(6,rate=lambda,lower.tail=FALSE)
s <- 4; t <- 3
cond <- pexp(s+t,rate=lambda,lower.tail=FALSE)/pexp(s,rate=lambda,lower.tail=FALSE)
directa <- pexp(t,rate=lambda,lower.tail=FALSE)
c(condicional=cond,directa=directa)
```

## Normal y estandarización


```r
mu <- 100; sigma <- 15
pnorm(120,mu,sigma)
pnorm(115,mu,sigma)-pnorm(85,mu,sigma)
qnorm(0.95,mu,sigma)
sapply(1:3,function(k) pnorm(k)-pnorm(-k))
```

## Simulación


```r
set.seed(2026)
u <- runif(50000,2,8)
e <- rexp(50000,0.25)
n <- rnorm(50000,100,15)
rbind(uniforme=c(mean(u),var(u)),exponencial=c(mean(e),var(e)),normal=c(mean(n),var(n)))
```
