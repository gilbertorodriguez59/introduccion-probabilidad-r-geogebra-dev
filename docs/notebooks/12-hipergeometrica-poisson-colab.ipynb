# Capítulo 12. Hipergeométrica y Poisson
Cuaderno complementario en R.


## Hipergeométrica y comparación binomial


```r
N <- 50; K <- 6; n <- 8
x <- 0:n
h <- dhyper(x, K, N-K, n)
b <- dbinom(x, n, K/N)
data.frame(x, hipergeometrica=h, binomial=b)
```

## Poisson y aproximación binomial


```r
n <- 1000; p <- 0.002; lambda <- n*p
x <- 0:8
data.frame(x, binomial=dbinom(x,n,p), poisson=dpois(x,lambda))
```

## Simulación


```r
set.seed(2026)
xh <- rhyper(50000,6,44,8)
xp <- rpois(50000,4)
c(media_h=mean(xh), var_h=var(xh), media_p=mean(xp), var_p=var(xp))
```
